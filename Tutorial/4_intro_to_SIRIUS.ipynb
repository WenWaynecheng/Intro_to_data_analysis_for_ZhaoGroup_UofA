# SIRUIS



SIRUIS is a free to use software used for small molecules and use MS2 to build a tree or network of peaks observed in the data. 

CSI:FingerID (bundled with SIRIUS) goes one step further: it takes that fragmentation tree and predicts a molecular fingerprint via a trained model, then searches a structure database (PubChem by default) for compounds with the closest predicted fingerprint. This gives ranked structure candidates even with zero reference spectra for that exact compound


you can download from their github page: https://github.com/sirius-ms/sirius



similar to mzmine, you do need an account to perform the jobs in the software

### How it works



https://v6.docs.sirius-ms.io/methods-background/#SIRIUS-molecular-formula

### data format


in MZMine or other processing software, the export module usually supports writing an output file  (containing MS1 and MS2) for SIRIUS (.mgf or.sirius file).

you may also just dump pre-procesing data (as .mzml file) either with vcvendor software or MSConvert (see `Tutorial > 1_MSConvert`)

<Br><br>
note that SIRIUS __requires__ high resolution data, and thus low resolution instrument like Ion Trap or Triple Quad is not accepted. 


you may simply drop the files (either from exported SIRIUS file or mzml) to the left of the main page.


for mzml file import, you will see the LC-MS tab on the top right where if you uses .mgf file this will not open (As it's processed already)


### how to use

you simply have to drop the file, check the filter on the left, and press compute all buttom on the top 


this will bring out the window of settings.  



one thing you need to change is the elemental composition. by defaul it uses elements that are usually not found in biomass burning (such as metals or halogens). you may exclude those by making the limit 0 to 0



<br>

note that the processing time can be long. you may check the progress with the `Jobs`


### Formula View




#### formula candidate list

the list on the top which you can see different formula/adducts and the score (which is sort by SOROIS score)

the green highlighted formula if tthe best candidate structure found by CSI:FingerID. note that this does not have to be the best SIURIUS score row. 


Sirius Score is a combination of the score from the isotope pattern analysis (Isotope Score) of the MS1 data and the fragmentation tree score (Tree Score) from the MS2 data.


#### mass spectra

the bottum left is the Mass spectra, by default the merged MS2 (if any). 

the blue sticks are annoatated by the fragmentation tree. SIRIUS by default treat those not annotated by the tree as noise

#### Fragmentation tree


Fragmentation trees are computed directly from the fragmentation spectrum without the need for spectral libraries or molecular structure databases.

the graph will show you the hypotehtical reaction. 


see monre on the tree here: https://v6.docs.sirius-ms.io/methods-background/#fragmentation-trees

### compound class through CANOPUS



this can be potentially useful for filtering peaks in the data

note that this class prediction doesnt rely on database but simply from the fragmentation tree. 

however, this is not detail compound class like coumarin class that's used in some databases. 


the publication here: [10.1038/s41587-020-0740-8](https://doi.org/10.1038/s41587-020-0740-8?urlappend=%3Futm_source%3Dresearchgate.net%26utm_medium%3Darticle)

### Structure view



this tab displays the candidate structures for the selected formula (as ranked by CSI:FingerID search score with highest-scoring candidate is highlighted in green). 


though more useful for metabolomics, you may filter by database.


for each candidate, there are color blocks on the top middle under __substructures__. 
- green block means the substructure is predicted by CSI:FiugerID and present in the feature. the size represent the reliability of the predictor
- the pink block repesent the substructures that are predicted to be __absent__ but found in the structure.

you may click the block and the corresponding structure will be highlighed on the left

### Substructure annotation view



the same table from strucutre view is on the top

the spectra on the bottum left shows the MS2 spectra
- black peaks are not used to explain the formula and not part of the fragmentation tree
- blue peaks are part of the fragmentation tree but no substructure
- green peaks are used for formula and has substructure
