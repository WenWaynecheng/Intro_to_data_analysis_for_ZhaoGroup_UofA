| Method | Advantages | Disadvantages | Best Used When |
|:---|:---|:---|:---|
| Centering | Removes baseline offset; preserves original data scale and units without modifying variance. | High-abundance or high-variance metabolites completely dominate multivariate models (e.g., PCA). | All variables share identical units and scales, and high-abundance features are expected to drive the primary biological signal. |
| Autoscaling (Standardization / Z-score) | Gives every variable equal weight (variance = 1); completely removes concentration scale bias. | Inflates technical noise, baseline ripple, and analytical artifacts in low-abundance features. | Unbiased exploratory analysis (e.g., PCA) across features spanning multiple orders of magnitude where low-abundance features carry equal biological weight |
| Pareto Scaling | Dampens the dominance of large peaks while remaining closer to raw data scales; inflates noise less than autoscaling | Does not fully eliminate scale dependence or the impact of extreme fold-changes | High-abundance peaks dominate the raw signal, but low-abundance features contain noticeable background/analytical noise |
| Range Scaling | Scales data relative to the biological range (Max−Min); removes concentration magnitude dependency | Highly sensitive to measurement outliers and single-point analytical errors | Biological min/max range bounds are meaningful, and the dataset has been strictly quality-checked for single-point outliers |
| Level Scaling | Focuses directly on relative percentage changes against the mean response | Severely inflates noise for features with near-zero means or high relative standard deviations | Biomarker discovery applications where percentage response or fold-change matters more than absolute concentration variance |
| Vast Scaling | Emphasizes metabolites with small relative fluctuations; incorporates group-specific variance stability | Ineffective for unguided/exploratory data lacking predefined sample groups or containing large induced variation | Targeted studies aiming to detect subtle shifts in normally tightly regulated baseline or housekeeping metabolites across defined groups |
| Log Transformation | Corrects heteroscedasticity (variance increasing with mean); converts multiplicative effects to additive ones | Undefined for zero values (requires artificial offset); can over-emphasize low values with high relative noise. | Data displays strong heteroscedasticity or log-normal distribution, and missing/zero values have been imputed or excluded |
| Power Transformation (sqrt(x)) | Reduces heteroscedasticity; handles true zero values natively without requiring artificial offsets | Weaker variance-stabilizing effect than log transformation; choice of power (e.g., square root) is arbitrary | Data exhibits heteroscedastic noise but contains genuine zero values that prevent direct log transformation |

as the paper suggests, there is no single "one-size-fits-all" pretreatment method. Researchers must select a method based on:

1. The specific biological question being asked.

2. The noise level and presence of outliers in the dataset.

3. The requirements of the chosen downstream statistical tool.
