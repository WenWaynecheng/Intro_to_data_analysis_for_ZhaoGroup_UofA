# Reading data 


we can read data from csv files 

```python
import os
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
current_path = os.getcwd()

file_path = os.path.join(current_path, "Sample data", "tube_furnace_positive_mode.csv")

# usually we use pd.read_csv or read_excel to get the data from csv or xlsx file
# csv is usually preferred since its light and easy to handle

# xlsx, on the other hand, usually require more detail handling such as which sheet you want 
# xlsx is also a bit harder to write due to formating but useful if  you want to automate the process

df = pd.read_csv(file_path)


# the first thing you get a data is to look at  it right away
# in notebook, if you simply have the df (or the variabel name of your data) at the last line
# notebook will try to show it

# but if your data has thousand of rows, we can check the first few row by df.head()
# or the last few rows by df.tail()
# by default it will show you 5 rows unless you specify it by entering the number of rows you want in the ()

df.head()
```
