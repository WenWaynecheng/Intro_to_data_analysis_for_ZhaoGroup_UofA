
# Data Conversion




https://proteowizard.sourceforge.io/

this is mostly the "default" tool for data conversion from the raw instrument file (e.g., `.raw` from Thermo, `.d` from Agilent or Bruker)

but we only use the side sofeware came with it called  **MSConvert** which you have to search it after installation



1. Selecting files



First thing is selecting your files with the **Browse** bottom on the top left corner. you can choose multiple files. 

you may also choose the **Output Directory** to wherever you wanted. Personally I keep the raw file in `0_Raw` folder and put the converted file in `1_Converted data` folder. just easy to track things 





2. Convertion options



Next you want to choose what to convert. usually we use mzML as the universal file format to process (which is readable for most programs).

you dont have to change anything here unless you have a specific use

3. Filters 



This is generic more toward what you're doing which you may filter out things or pre-process the files. 



Peak picking usually is the one recommanded to use. Though there's are two options : **Vendor** and the **CWT** (continuous wavelet transform). 

The vendor  means  it's using the instrument manufacturer's peroperties. **For orbitrap, we can use this as it is.**

CWT is another algorithm that search better for tiny peaks. this may have more features then using Vendor option. though we dont really know until we try both. 

If you want to know more about CWT, they published a paper in 2014: https://doi.org/10.1021/pr500886y



4. Run it


a window will pop up for the progress. close it after everything is finished with green bar for each file.
